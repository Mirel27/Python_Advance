```python
import requests
import json
```

```python
## Get api call

url="https://jsonplaceholder.typicode.com/posts/1"

response=requests.get(url)

# Check the status code of the response
response.status_code
response.url
response.headers

response.text

response.json()  # Convert the response to JSON format
```

```python
#Parsed json data
data=response.json()
data = json.dumps(data, indent=4)  # Convert the JSON data to a formatted string
```
