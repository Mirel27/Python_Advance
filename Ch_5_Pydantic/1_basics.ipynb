```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    id:int
    name: str
    age: int
    department: str
```

```python
obj1 = employee_schema(id=1, name='Mirel', age=30, department='IT')
obj1
```

```python
obj = employee_schema(**{
    "id": 1,
    "name": "Mirel",
    "age": 25,
    "department": "IT",
})
obj
```

## Validation Error because of data drift in the schema


```python
obj1 = employee_schema(**{
    "id": 1,
    "name": "Mirel",
    "age": 30,
    "department": 12
})
obj1 #ValidationError: 1 validation error for employee_schema
```

## COERCION


```python
obj1 = employee_schema(**{
    "id": "123",
    "name": "Mirel",
    "age": 30,
    "department": "ID"
})
obj1 #No error because pydantic will try to convert the string to int as it is not strict 
#
```

## Strict Types


```python
from pydantic import  StrictInt

class employee_schema(BaseModel):
    id:StrictInt
    name: str
    age: int
    department: str
```

```python
obj4 = employee_schema(**{
    "id": "123",    
    "name": "Mirel",
    "age": 30,
    "department": "ID"
})

obj4 #ValidationError: 1 validation error for employee_schema
```
