```python
from pydantic import BaseModel, Field, StrictInt
from pydantic import EmailStr
from typing import Optional,Literal
```

```python
class Employee(BaseModel):
    id: StrictInt
    name: str = Field(..., description="The name of the employee",min_length=1, max_length=50)
    age: int = Field(..., description="The age of the employee", gt=18 , lt = 65)
    department: str = Field(..., description="The department of the employee", min_length=2, max_length=100)
    email: EmailStr = Field(..., description="The email of the employee")
    department: Optional[str] = None
    gender: Literal['Male', 'Female', 'Other'] = Field(..., description="The gender of the employee")
```

```python
obj5 = Employee(**{
    "id": 1,
    "name": "Mirel",
    "age": 30,
    "department": "IT",
    "email": "mae@gmail.com",
    "gender": "Male",
})  
obj5   
```
